```python
import pandas as pd
from sqlalchemy import create_engine


DB_USER = "postgres"
DB_PASSWORD = "sql123"
DB_HOST = "localhost"
DB_PORT = "5433"
DB_NAME = "superstore_db"


engine = create_engine(f"postgresql://{DB_USER}:{DB_PASSWORD}@{DB_HOST}:{DB_PORT}/{DB_NAME}")


table_name = "superstore_orders"

df.to_sql(
    name=table_name,
    con=engine,
    if_exists="replace",  
    index=False,          
    chunksize=5000      
)

print(f"Successfully uploaded {len(df)} rows to table '{table_name}' in PostgreSQL!")
```
